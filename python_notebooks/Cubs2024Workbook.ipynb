```python
import pandas as pd
import altair as alt
#disable max row
alt.data_transformers.disable_max_rows()

with open('retrosheet_data/2024CHN.EVN', 'r') as f:
    lines = f.readlines()
#lists
game_ids = []
daynight_flags = []
temps = []
attendances = []

for line in lines:
    #gameid
    if line.startswith('id,'):
        game_ids.append(line.strip().split(',')[1])
        #daynight
    elif line.startswith('info,daynight,'):
        daynight_flags.append(line.strip().split(',')[2])
        #temp
    elif line.startswith('info,temp,'):
        temps.append(int(line.strip().split(',')[2]))
        #attendance
    elif line.startswith('info,attendance,'):
        attendances.append(int(line.strip().split(',')[2]))

min_len = min(len(game_ids), len(daynight_flags), len(temps), len(attendances))
#dataframe
games_df = pd.DataFrame({
    'game_id': game_ids[:min_len],
    'daynight': [x.lower() for x in daynight_flags[:min_len]],
    'temp': temps[:min_len],
    'attendance': attendances[:min_len]
})

df = games_df.copy()
#dayvsnight bar plot
driver_plot = alt.Chart(df).mark_bar().encode(
    x=alt.X('daynight:N', title='Game Type'),
    y=alt.Y('count():Q', title='Number of Games'),
    color=alt.Color('daynight:N', scale=alt.Scale(domain=['day', 'night'], range=['orange', 'midnightblue'])),
    tooltip=['daynight:N', 'count():Q']
).properties(
    title='Cubs 2024: Day vs Night Games'
).add_params(
    alt.selection_point(fields=['daynight'], bind='legend')
).interactive()
#scatter plot att vs. temp
attendance_temp_plot = alt.Chart(df).mark_point(filled=True, size=60).encode(
    x=alt.X('attendance:Q', title='Attendance'),
    y=alt.Y('temp:Q', title='Temperature'),
    color=alt.Color('daynight:N', scale=alt.Scale(domain=['day', 'night'], range=['orange', 'midnightblue'])),
    tooltip=['attendance', 'temp', 'daynight']
).properties(
    title='Attendance vs Temperature for Cubs 2024 Games'
).add_params(
    alt.selection_point(fields=['daynight'], bind='legend')
).interactive()
attendance_temp_plot | driver_plot
driver_plot.save('driver_plot.html')
```
