```python
import pandas as pd
url = 'https://raw.githubusercontent.com/UIUC-iSchool-DataViz/is445_data/main/bfro_reports_fall2022.csv'
df = pd.read_csv(url)
df.head()
```

```python
state_counts = df['state'].value_counts().reset_index()
state_counts.columns = ['state', 'count']
state_counts.head()
```

```python
import altair as alt
bar_chart = alt.Chart(state_counts).mark_bar().encode(
    x=alt.X('count:Q', title='Number of Sightings'), y=alt.Y('state:N', sort='-x', title='State'),
    color=alt.Color('count:Q', scale=alt.Scale(scheme='reds'))
).properties(
    title='Bigfoot Sightings by State',
    width=600, height=600
)
bar_chart
```

```python
df['date'] = pd.to_datetime(df['date'], errors='coerce')
df['year'] = df['date'].dt.year

yearly_counts = df['year'].value_counts().sort_index().reset_index()
yearly_counts.columns = ['year', 'count']
```

```python
bar_plot = alt.Chart(yearly_counts).mark_bar().encode(
    x='year:O', y='count:Q',
    color=alt.Color('count:Q', scale=alt.Scale(scheme='Blues'))
).properties(
    title='Bigfoot Sighting by Year',
    width=600, height=600
)
bar_plot
```

```python
bar_plot = alt.Chart(yearly_counts).mark_bar().encode(
    x='year:O', y='count:Q',
    color=alt.Color('count:Q', scale=alt.Scale(scheme='blues')),
    tooltip=['year', 'count']
).properties(
    title='Bigfoot Sighting by Year',
    width=600, height=600
)
bar_plot
```
